Lists, Sets, Tuples, Dictionaries

```python
#1.LISTS:A list is an ordered, mutable collection that allows duplicates
fruits = ['apple', 'banana', 'cherry', 'apple']
print("Original list:", fruits)
print("First fruit:", fruits[0])# Accessing elements by index
fruits.append('orange')# Adding an item
print("After appending:", fruits)
fruits.remove('banana')# Removing an item
print("After removing banana:", fruits)
print("Number of fruits:", len(fruits))# List length
```

```python
#2.TUPLES:A tuple is an ordered, immutable collection that allows duplicates
dimensions = (10, 20, 30)
print("Tuple:", dimensions)
print("Second element:", dimensions[1])# Accessing elements
```

```python
#3.SETS:A set is an unordered, mutable collection with NO duplicates
unique_numbers = {1, 2, 3, 2, 1}
print("Set (no duplicates):", unique_numbers)
unique_numbers.add(4) #Adding elements
print("After adding 4:", unique_numbers)
unique_numbers.remove(2) #Removing elements
print("After removing 2:", unique_numbers)
#Set operations
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
```

```python
#4.DICTIONARIES:A dictionary stores key-value pairs (like a map or JSON object)
student = {
    'name': 'Alice',
    'age': 21,
    'course': 'Python'
}
print("Dictionary:", student)
print("Student name:", student['name']) # Access value by key
# Add or update value
student['grade'] = 'A'
print("After adding grade:", student)
# Remove a key
del student['age']
print("After deleting age:", student)
# Loop through dictionary
for key, value in student.items():
    print(f"{key}: {value}")
# Bonus: Dictionary from two lists using zip()
keys = ['id', 'name', 'marks']
values = [101, 'John', 88]
student_info = dict(zip(keys, values))
print("Student Info from zip:", student_info)
```
